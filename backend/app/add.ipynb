```python
import json
import requests
from pathlib import Path

# =====================================================
# ⚙️ CONFIGURATION
# =====================================================
API_URL = "http://127.0.0.1:8000/zones"  # ton endpoint FastAPI
REGION_ID = "15bfe49a-49e0-41bd-8c97-1b2071d7c707"
GEOJSON_PATH = Path(r"C:\Users\accent\Desktop\wattara\backend\data\geojsonBF.geojson")

# =====================================================
# 📦 Lecture du GeoJSON
# =====================================================
with open(GEOJSON_PATH, "r", encoding="utf-8") as f:
    data = json.load(f)

added = 0
skipped = 0

# =====================================================
# 🚀 Envoi des requêtes POST
# =====================================================
for feature in data.get("features", []):
    props = feature.get("properties", {})
    geom = feature.get("geometry", {})

    name = props.get("name")
    place = props.get("place")

    if not name:
        continue

    payload = {
        "name": name,
        "place": place,
        "geometry": geom,
        "region_id": REGION_ID,
        "color_code": "#%06x" % (hash(name) & 0xFFFFFF),
    }

    response = requests.post(API_URL, json=payload)
    if response.status_code == 200:
        added += 1
    elif response.status_code == 409:
        skipped += 1
    else:
        print(f"❌ Erreur pour {name}: {response.status_code} - {response.text}")

print(f"✅ Import terminé : {added} zones ajoutées, {skipped} ignorées.")
```
